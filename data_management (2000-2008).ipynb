**Internal Test**

```python
RNG = 42  # reproducibility

# --- Separate NA vs. complete cases ---
df_na     = df['2000'][df['2000'].isna().any(axis=1)]
df_non_na = df['2000'].dropna()

N = len(df['2000'])
target_train = int(round(0.8 * N))
target_val   = int(round(0.1 * N))
target_test  = N - target_train - target_val  # ensures totals sum to N

# --- All NA must be in train ---
train_from_na = df_na

# How many complete-case rows do we still need to hit the 80% target?
need_non_na_for_train = max(0, target_train - len(train_from_na))

# Shuffle complete cases
df_non_na_shuf = df_non_na.sample(frac=1.0, random_state=RNG)

# Take some complete cases into train to reach target (as much as possible)
train_from_non_na = df_non_na_shuf.iloc[:need_non_na_for_train]
rem_non_na = df_non_na_shuf.iloc[need_non_na_for_train:]

# Now split the remaining complete cases into val/test (aim for 10%/10%)
if len(rem_non_na) < (target_val + target_test):
    # Not enough complete cases to satisfy 10%/10% exactly.
    # Use as many as possible; split ~50/50 for val/test.
    print("[WARN] Not enough complete-case rows to meet exact 10%/10%. "
          "Allocating all remaining complete cases to val/test.")
    val_size  = len(rem_non_na) // 2
    test_size = len(rem_non_na) - val_size
else:
    val_size  = target_val
    test_size = target_test

val  = rem_non_na.iloc[:val_size]
test = rem_non_na.iloc[val_size: val_size + test_size]

# Build final train set: all NA + selected complete cases
train = pd.concat([train_from_na, train_from_non_na], axis=0)

# --- Sanity checks ---
assert train.index.is_unique and val.index.is_unique and test.index.is_unique
assert set(train.index).isdisjoint(val.index) and set(train.index).isdisjoint(test.index)
assert set(val.index).isdisjoint(test.index)
assert val.isna().sum().sum() == 0 and test.isna().sum().sum() == 0  # val/test no NA
# Train may contain NA by design

print("=== Split Summary (target 80/10/10) ===")
print(f"Total rows: {N}")
print(f"NA rows (all in train): {len(df_na)}")
print(f"Complete-case rows: {len(df_non_na)}")
print(f"Train size: {len(train)}  (NA in train: {train.isna().any(axis=1).sum()})")
print(f"Val size:   {len(val)}    (NA in val:   {val.isna().any(axis=1).sum()})")
print(f"Test size:  {len(test)}   (NA in test:  {test.isna().any(axis=1).sum()})")
```

```python
# --- Combine train + val ---
train_val = pd.concat([train, val], axis=0, ignore_index=True)

# --- Ensure output directory exists ---
out_dir = "data/2000-2008"
os.makedirs(out_dir, exist_ok=True)

# --- Save CSVs ---
train_val_path = os.path.join(out_dir, "2000-2008internal_trainval.csv")
test_path      = os.path.join(out_dir, "2000-2008internal_test.csv")

train_val.to_csv(train_val_path, index=False)
test.to_csv(test_path, index=False)

# --- save train/val separately for auditing ---
train.to_csv(os.path.join(out_dir, "2000-2008internal_train.csv"), index=False)
val.to_csv(os.path.join(out_dir, "2000-2008internal_val.csv"), index=False)

print("\n=== Internal CSVs saved ===")
print(f"Train+Val → {train_val_path} (rows: {len(train_val)})")
print(f"Test      → {test_path}      (rows: {len(test)})")

if 'y' in train_val.columns:
    print("\nTrain+Val label counts:\n", train_val['y'].value_counts())
if 'y' in test.columns:
    print("\nTest label counts:\n", test['y'].value_counts())
```

```python
for i in range(1, 11):
    seed = 42 + i

    # Shuffle complete-case pool each fold
    non_na_shuf = shuffle(df_non_na, random_state=seed)

    # Fill train up to the 80% target using complete cases (all NA already go to train)
    need_non_na_for_train = max(0, target_train - len(df_na))
    train_non_na = non_na_shuf.iloc[:need_non_na_for_train]
    rem_non_na   = non_na_shuf.iloc[need_non_na_for_train:]

    # Allocate remaining complete cases to val/test (aim 10%/10%)
    if len(rem_non_na) < (target_val + target_test):
        # If not enough complete cases to hit 10%+10% exactly, use what we have
        print(f"[Fold {i}] WARN: only {len(rem_non_na)} complete-case rows left "
              f"for val+test (need {target_val + target_test}). "
              "Allocating as many as possible.")
        val_size  = min(target_val, len(rem_non_na))
        test_size = min(target_test, len(rem_non_na) - val_size)
    else:
        val_size, test_size = target_val, target_test

    val_set  = rem_non_na.iloc[:val_size]
    test_set = rem_non_na.iloc[val_size: val_size + test_size]

    # Build final train (all NA + selected complete cases)
    train_set = pd.concat([df_na, train_non_na], axis=0)

    # --- Sanity checks (val/test must have no NA) ---
    assert val_set.isna().sum().sum() == 0,  f"[Fold {i}] Validation set has NA!"
    assert test_set.isna().sum().sum() == 0, f"[Fold {i}] Test set has NA!"

    # Optional: print class balance if you have a label column 'y'
    if 'y' in train_set.columns:
        print(f"[Fold {i}] Train y counts:\n{train_set['y'].value_counts()}\n")

    # Save splits
    train_set.to_csv(f'{out_dir}/2000-2008int_train_{i}.csv', index=False)
    val_set.to_csv(  f'{out_dir}/2000-2008int_val_{i}.csv',   index=False)
    test_set.to_csv( f'{out_dir}/2000-2008int_test_{i}.csv',  index=False)

    # Brief summary
    print(f"[Fold {i}] sizes → train:{len(train_set)}  val:{len(val_set)}  test:{len(test_set)}  "
          f"(NA in train: {train_set.isna().any(axis=1).sum()})")
```

**time, event for COX model**

```python
def first_1_else_last_0_with_value(row,year=2000):
    ones = row[row == 1]
    if not ones.empty:
        if ones.index[0] == None:
            col = ones.index[0]
        else:
            col = int(ones.index[0][:2]) - (year-2000)
        return pd.Series([col, 1])

    zeros = row[row == 0]
    if not zeros.empty:
        if zeros.index[-1] == None:
            col = zeros.index[-1]
        else:
            col = int(zeros.index[-1][:2]) - (year-2000)
        return pd.Series([col, 0])

    return pd.Series([np.nan, np.nan])
```
