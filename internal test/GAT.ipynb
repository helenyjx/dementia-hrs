#### Internal Test with stratified data splitting via miceforest imputation

```python
df_full = pd.read_csv('../data/preprocessed data/2000(00-06y)encode.csv')
X_full, y_full = impute_data_and_y(df_full, 'mice', X_columns, y_column)
```

```python
train_sets = []
train_val_sets = []
val_sets = []
test_sets = []
for i in range(10):
    # make train+val and test data split 8:2
    X_train_val, X_test, y_train_val, y_test = train_test_split(X_full, y_full, test_size=0.2, stratify=y_full, random_state=i)
    train_val_sets.append((X_train_val, y_train_val))
    test_sets.append((X_test, y_test))
    # make train and val data split 6:2. finally train:val:test = 6:2:2
    X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=0.25, stratify=y_train_val, random_state=i)
    train_sets.append((X_train, y_train))
    val_sets.append((X_val, y_val))
```

```python
candidate_results = {}
for train, val in zip(train_sets,val_sets):
    X_train, y_train = train
    X_val, y_val = val
    combinations = itertools.product(*param_grid.values())
    for parameter in combinations:
        candidate = list(param_grid.keys())
        try:
            m = GATClassifier(**dict(zip(candidate, parameter)))
            m.fit(X_train, y_train)
            y_val_prob = m.predict_proba(X_val)[:, 1]
            aupr = average_precision_score(y_val, y_val_prob)
            candidate_results.setdefault(parameter, []).append(aupr)
        except Exception:
            continue
avg_val_aupr = {cand: np.mean(scores) for cand, scores in candidate_results.items()}
best_candidate = max(avg_val_aupr, key=avg_val_aupr.get)
best_val_aupr = avg_val_aupr[best_candidate]
print(
    f"For imputation method 'mice', best candidate: {best_candidate} "
    f"with average validation AUPR = {pct(best_val_aupr)}"
        )
```

```python
per_fold = []
optimal_thresholds_list = []

# Step 1: Find optimal threshold for each fold
for train_val, test in zip(train_val_sets, test_sets):
    X_train_val, y_train_val = train_val
    X_test, y_test = test
    final_model = GATClassifier(**dict(zip(candidate, best_candidate)))
    final_model.fit(X_train_val, y_train_val)
    y_prob = final_model.predict_proba(X_test)[:, 1]
    
    # Find optimal threshold based on F1 score
    precision, recall, thresholds = precision_recall_curve(y_test, y_prob)
    f1_scores = 2 * (precision * recall) / (precision + recall + 1e-10)
    optimal_idx = np.argmax(f1_scores)
    optimal_threshold = thresholds[optimal_idx] if optimal_idx < len(thresholds) else 0.5
    
    optimal_thresholds_list.append(optimal_threshold)

# Step 2: Use mean optimal threshold for all folds
mean_optimal_threshold = np.mean(optimal_thresholds_list)
print(f"Mean Optimal Threshold: {mean_optimal_threshold:.4f}")
print(f"Optimal thresholds per fold: {[f'{t:.4f}' for t in optimal_thresholds_list]}")

# Step 3: Calculate metrics using optimal threshold
per_fold = []
for train_val, test in zip(train_val_sets, test_sets):
    X_train_val, y_train_val = train_val
    X_test, y_test = test
    final_model = GATClassifier(**dict(zip(candidate, best_candidate)))
    final_model.fit(X_train_val, y_train_val)
    y_prob = final_model.predict_proba(X_test)[:, 1]
    y_pred = (y_prob >= mean_optimal_threshold).astype(int)
    
    per_fold.append({
        "AUC": roc_auc_score(y_test, y_prob),
        "ACC": accuracy_score(y_test, y_pred),
        "AUPR": average_precision_score(y_test, y_prob),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1": f1_score(y_test, y_pred),
    })

# Step 4: Calculate means and stds across folds
aucs      = np.array([m["AUC"]       for m in per_fold])
accs      = np.array([m["ACC"]       for m in per_fold])
auprs     = np.array([m["AUPR"]      for m in per_fold])
precisions = np.array([m["Precision"] for m in per_fold])
recalls   = np.array([m["Recall"]    for m in per_fold])
f1s       = np.array([m["F1"]        for m in per_fold])

mean_auc,  std_auc  = aucs.mean(),  aucs.std()
mean_acc,  std_acc  = accs.mean(),  accs.std()
mean_aupr, std_aupr = auprs.mean(), auprs.std()
mean_precision, std_precision = precisions.mean(), precisions.std()
mean_recall,    std_recall    = recalls.mean(),    recalls.std()
mean_f1,        std_f1        = f1s.mean(),        f1s.std()

print("\nGATClassifier with MICE Imputation — Internal Test")
print(f"Optimal Threshold Used: {mean_optimal_threshold:.4f}")
print(
    f"Final test metrics: "
    f"AUC = {pct(mean_auc)}, ACC = {pct(mean_acc)}, AUPR = {pct(mean_aupr)}, "
    f"Precision = {pct(mean_precision)}, Recall = {pct(mean_recall)}, F1 = {pct(mean_f1)}"
)
print(
    f"Standard deviations across folds: "
    f"AUC_std = {pct(std_auc)}, ACC_std = {pct(std_acc)}, AUPR_std = {pct(std_aupr)}, "
    f"Precision_std = {pct(std_precision)}, Recall_std = {pct(std_recall)}, F1_std = {pct(std_f1)}"
)
```

```python
def permutation_feature_importance(model, X, y, metric='roc_auc', n_repeats=10, random_state=None):
    """
    Calculate permutation feature importance for any model.
    
    Parameters:
    -----------
    model : trained model
        Must have predict() or predict_proba() method
    X : array-like or dict
        Feature data. Can be numpy array, pandas DataFrame, or dict with node features
    y : array-like
        True labels
    metric : str or callable
        Metric to use: 'roc_auc', 'aupr', 'accuracy', 'f1', or custom function
    n_repeats : int
        Number of times to permute each feature
    random_state : int or None
        Random seed for reproducibility
        
    Returns:
    --------
    importances_mean : array
        Mean importance for each feature
    importances_std : array
        Standard deviation of importance for each feature
    feature_names : list
        List of feature names
    """
    rng = np.random.RandomState(random_state)
    
    # Handle different input types
    if isinstance(X, dict):
        # For graph data with node features
        X_array = X['node_features']
        n_features = X_array.shape[1]
        feature_names = [f'feature_{i}' for i in range(n_features)]
    elif isinstance(X, pd.DataFrame):
        X_array = X.values
        n_features = X.shape[1]
        feature_names = X.columns.tolist()
    else:
        X_array = np.array(X)
        n_features = X_array.shape[1]
        feature_names = [f'feature_{i}' for i in range(n_features)]
    
    # Define metric function
    if metric == 'roc_auc':
        if hasattr(model, 'predict_proba'):
            baseline_score = roc_auc_score(y, model.predict_proba(X)[:, 1])
            metric_fn = lambda y_true, y_pred: roc_auc_score(y_true, y_pred[:, 1])
            use_proba = True
        else:
            raise ValueError("Model must have predict_proba for ROC-AUC")
    elif metric == 'aupr' or metric == 'auprc':
        if hasattr(model, 'predict_proba'):
            baseline_score = average_precision_score(y, model.predict_proba(X)[:, 1])
            metric_fn = lambda y_true, y_pred: average_precision_score(y_true, y_pred[:, 1])
            use_proba = True
        else:
            raise ValueError("Model must have predict_proba for AUPR")
    elif metric == 'accuracy':
        baseline_score = accuracy_score(y, model.predict(X))
        metric_fn = accuracy_score
        use_proba = False
    elif metric == 'f1':
        baseline_score = f1_score(y, model.predict(X), average='binary')
        metric_fn = lambda y_true, y_pred: f1_score(y_true, y_pred, average='binary')
        use_proba = False
    elif callable(metric):
        baseline_score = metric(y, model.predict(X))
        metric_fn = metric
        use_proba = False
    else:
        raise ValueError(f"Unknown metric: {metric}")
    
    print(f"Baseline {metric} score: {baseline_score:.4f}")
    
    # Store importance scores
    importances = np.zeros((n_features, n_repeats))
    
    # Calculate importance for each feature
    for feat_idx in range(n_features):
        for repeat in range(n_repeats):
            # Create a copy of the data
            if isinstance(X, dict):
                X_permuted = {k: v.copy() if isinstance(v, np.ndarray) else v 
                             for k, v in X.items()}
                X_permuted_array = X_permuted['node_features']
            else:
                X_permuted = X.copy() if isinstance(X, pd.DataFrame) else X_array.copy()
                X_permuted_array = X_permuted.values if isinstance(X_permuted, pd.DataFrame) else X_permuted
            
            # Permute the feature
            X_permuted_array[:, feat_idx] = rng.permutation(X_permuted_array[:, feat_idx])
            
            # Update the dict if needed
            if isinstance(X, dict):
                X_permuted['node_features'] = X_permuted_array
            else:
                if isinstance(X, pd.DataFrame):
                    X_permuted.iloc[:, feat_idx] = X_permuted_array[:, feat_idx]
                else:
                    X_permuted = X_permuted_array
            
            # Calculate score with permuted feature
            if use_proba:
                y_pred = model.predict_proba(X_permuted)
            else:
                y_pred = model.predict(X_permuted)
            
            permuted_score = metric_fn(y, y_pred)
            
            # Importance is the decrease in performance
            importances[feat_idx, repeat] = baseline_score - permuted_score
    
    # Calculate mean and std
    importances_mean = importances.mean(axis=1)
    importances_std = importances.std(axis=1)
    
    return importances_mean, importances_std, feature_names


def plot_feature_importance(importances_mean, importances_std, feature_names, 
                           top_n=None, figsize=(10, 8), title=None):
    """
    Plot feature importances with error bars.
    """
    import matplotlib.pyplot as plt
    
    # Sort by importance
    indices = np.argsort(importances_mean)[::-1]
    
    if top_n is not None:
        indices = indices[:top_n]
    
    sorted_names = [feature_names[i] for i in indices]
    sorted_importances = importances_mean[indices]
    sorted_std = importances_std[indices]
    
    plt.figure(figsize=figsize)
    plt.barh(range(len(sorted_names)), sorted_importances, xerr=sorted_std)
    plt.yticks(range(len(sorted_names)), sorted_names)
    plt.xlabel('Permutation Importance')
    plt.ylabel('Features')
    
    if title is None:
        plt.title('Feature Importance (with std dev)')
    else:
        plt.title(title)
    
    plt.tight_layout()
    plt.gca().invert_yaxis()
    
    return plt.gcf()

importances_mean, importances_std, feature_names = permutation_feature_importance(
    final_model, X_test, y_test, metric="aupr", n_repeats=10
)
```
